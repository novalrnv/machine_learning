# Lab 3 - Klasifikasi Naive Bayes dengan Data Multimoda (Teks)

Mengikuti langkah-langkah pada modul JS05 - Klasifikasi 1, Lab 3.

Kita akan menggunakan data `spam.csv` yang berisi data teks SMS dengan label **spam** dan **ham**. Spam adalah SMS sampah, sedangkan ham adalah sebaliknya.

## Langkah 1 - Load Data

```python
import numpy as np
import pandas as pd

df = pd.read_csv('spam.csv', encoding='latin-1') # spesifikasi encoding diperlukan karena data tidak menggunakan UTF-8

df.head()
```

## Langkah 2 - Preprocessing

### Langkah 2a - Drop Kolom

```python
# Drop 3 kolom terakhir dengan fungsi iloc
df = df.drop(df.iloc[:,2:], axis=1)

# Ubah nama kolom v1 dan v2
df.columns = ['Labels', 'SMS']

# Cek data
df.head()
```

### Langkah 2b - Inspeksi Data

```python
# Cek Jumlah Data Per Kelas
print(df['Labels'].value_counts())
print('\n')

# Cek Kelengkapan Data
print(df.info())
print('\n')

# Cek Statistik Deskriptif
print(df.describe())
```

### Langkah 2c - Encoding Label

```python
# Data untuk label
new_labels = {
    'spam': 1,
    'ham': 0
}

# Encode label
df['Labels'] = df['Labels'].map(new_labels)

# Cek data
df.head()
```

### Langkah 2d - Pisahkan Fitur dengan Label

```python
X = df['SMS'].values
y = df['Labels'].values
```

## Langkah 3 - Ekstraksi Fitur

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer

# Split data training dan testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Inisiasi CountVectorizer
bow = CountVectorizer()

# Fitting dan transform X_train dengan CountVectorizer
X_train = bow.fit_transform(X_train)

# Transform X_test
# Mengapa hanya transform? Supaya model tidak mengetahui parameter yang digunakan
# oleh CountVectorizer untuk fitting data X_train, sehingga data testing
# dapat tetap menjadi data yang asing bagi model nantinya
X_test = bow.transform(X_test)
```

```python
print(len(bow.get_feature_names_out()))
print(f'Dimensi data: {X_train.shape}')
```

## Langkah 4 - Training dan Evaluasi Model

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Inisiasi MultinomialNB
mnb = MultinomialNB()

# Fit model
mnb.fit(X_train, y_train)

# Prediksi dengan data training
y_pred_train = mnb.predict(X_train)

# Evaluasi akurasi data training
acc_train = accuracy_score(y_train, y_pred_train)

# Prediksi dengan data testing
y_pred_test = mnb.predict(X_test)

# Evaluasi akurasi data testing
acc_test = accuracy_score(y_test, y_pred_test)

# Print hasil evaluasi
print(f'Hasil akurasi data train: {acc_train}')
print(f'Hasil akurasi data test: {acc_test}')
```
