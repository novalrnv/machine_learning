##Deskripsi Tugas

Pada tugas pratikum ini Anda akan menggunakan data "Wisconsin Breast Cancer". Data tersebut terdiri dari 569 data yang digunakan untuk mendiagnonis jenis kanker Malignant (M) dan Benign (B). Tugas Anda adalah,

1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

2. Lakukan proses encoding pada kolom "diagnosis".

3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv('/content/wbc.csv')
df.head()
```

```python
# Cek kolom yang tidak dapat digunakan
print("Jumlah nilai unik 'id':", df['id'].nunique(), "dari", len(df), "baris (bersifat identitas, bukan fitur)")
print("Jumlah nilai kosong 'Unnamed: 32':", df['Unnamed: 32'].isnull().sum(), "dari", len(df), "baris")

# Buang kolom yang tidak dapat digunakan
drop_cols = ['id', 'Unnamed: 32']
df_clean = df.drop(columns=drop_cols)

print("\nVariabel yang tidak dapat digunakan:", drop_cols)
print("Variabel yang dapat digunakan:", list(df_clean.columns))
print("Shape setelah pemisahan:", df_clean.shape)
```

##1. Pemisahan variabel:
id (identitas unik, tidak relevan) dan Unnamed: 32 (kolom kosong semua NaN) dibuang. diagnosis sebagai target, 30 kolom sisanya sebagai fitur.

```python
le = LabelEncoder()
df_clean['diagnosis'] = le.fit_transform(df_clean['diagnosis'])

print("Mapping encoding:", dict(zip(le.classes_, le.transform(le.classes_))))
print()
print(df_clean['diagnosis'].value_counts())
```

##2. Encoding diagnosis
pakai LabelEncoder: B (Benign) > 0, M (Malignant) > 1.

```python
feature_cols = df_clean.columns.drop('diagnosis')

scaler = StandardScaler()
df_scaled = df_clean.copy()
df_scaled[feature_cols] = scaler.fit_transform(df_clean[feature_cols])

df_scaled.head()
```

pakai StandardScaler pada 30 kolom fitur numerik (target tidak ikut di-scale), hasilnya rata-rata = 0 dan std = 1.
